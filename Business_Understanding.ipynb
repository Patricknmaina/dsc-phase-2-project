### BUSINESS PROBLEM
Your company now sees all the big companies creating original video content and they want to get in on the fun. They have decided to create a new movie studio, but they don’t know anything about creating movies. You are charged with exploring what types of films are currently doing the best at the box office. You must then translate those findings into actionable insights that the head of your company's new movie studio can use to help decide what type of films to create.


### BUSINESS UNDERSTANDING

The company's objective is to enter the entertainment industry by establishing a new movie studio. The goal is to create films that will intrigue the audiences and perform well at the box office. Success will depend on identifying key market trends and using them to inform the studio's production plan.

### Objectives

1. To identify the film genres, production budgets, and release periods most strongly associated with box office success.
2. To develop data-driven recommendations for genre selection, and release timing for new film projects.
3. To support strategic decision-making for the new movie studio by providing actionable insights based on historical box office data.

### PROJECT PLAN

### Overview/Background
The movie industry produces alot of data, but using this information effectively remains a challenge. By doing an analysis of movie datasets from IMDb, Rotten Tomatoes, BoM AND TMDB this project seeks to come up with patterns and data that will lead to strategic decision-making. This analysis will evaluate how factors such as audience ratings, genres and regional preferences, can influence popularity and profitability. The findings will assist stakeholders in understanding the dynamics of the movie market.


### Challenges
Integrating data from diverse sources poses some problems such as handling missing values, dealing with inconsistencies and merging some datasets. Additionally, coming up with actionable insights requires robust analysis techniques so as to manage exploratory and predictive approaches. 


### Proposed solution
The project will use the CRISP-DM technique to deal with these challenges systematically. The data will be rigorously cleaned and intergrated  to create one dataset that can be used to come up with stellar recommendations.

### Conclusion
This project aims to enable stakeholders to optimize their strategies in production, marketing and distribution through the transformation of the raw movie data into meaningful insights. This structured approach ensures the analysis is reliable and actionable, so as to help in making data-driven decisions that enhance profitability, audience engagement and popularity of the studio.
